**¿Qúe es la tuberculosis?**
Hoy
que es la tuberculosis
La tuberculosis es una enfermedad infecciosa causada por la bacteria Mycobacterium tuberculosis, que afecta principalmente a los pulmones, aunque también puede dañar otros órganos como riñones, huesos o cerebro. Se transmite por vía aérea cuando una persona enferma tose, estornuda o habla, y aunque es grave, es prevenible y curable con tratamiento adecuado.

**🔍 Métodos principales de diagnóstico**
1. Pruebas iniciales
Prueba cutánea de la tuberculina (Mantoux): Se inyecta una pequeña cantidad de proteína derivada de la bacteria bajo la piel. Una reacción positiva indica infección, pero no distingue entre TB activa y latente.

Prueba de sangre (IGRA): Detecta la respuesta inmunitaria frente a Mycobacterium tuberculosis. Más precisa en personas vacunadas con BCG.

2. Estudios de imagen
Radiografía de tórax: Permite observar lesiones pulmonares típicas de la TB activa.

3. Pruebas de laboratorio
Análisis de esputo: Se examina la flema bajo el microscopio para detectar bacilos.

Cultivo microbiológico: Considerado el “estándar de oro”, pero puede tardar de 2 a 8 semanas en dar resultados.

Pruebas moleculares (PCR): Detectan fragmentos de ADN bacteriano en pocas horas, aunque no siempre diferencian entre infección activa y pasada.


4. Diagnóstico avanzado
Pruebas de resistencia a medicamentos: Determinan si la bacteria es resistente a fármacos como la rifampicina o la isoniazida.

Biosensores fluorescentes (2026): Detectan la proteína MPT64, marcador específico de TB activa, en solo 60 minutos. Son rápidos, selectivos y de bajo coste, lo que los hace prometedores para países con recursos limitados.

**Explorar los valores nulos/faltanres/NA**

```python
# Cuantos valores faltantes hay por columna?
who.isna().sum()
```

```python
# Explorar los valores faltantes en la columna iso2
who[who.iso2.isna()]
```

```python
# Los valores NA se debe a que el iso2 de Namibia es "NA", y pandas lo toma como un valor faltante
who.loc[who.country == "Namibia", "iso2"] = "NA"
```

```python
# Las demas columnas de casos con valores faltantes van a tener un valor dafault de 0
who.fillna(0, inplace=True)
```

```python
# Cuantos valores faltantes hay por columna en la tbla de pop?
pop.isna().sum()
```

```python
# Como puedo convertir columnas en filas utilizando Pandas?:: Existe la función melt()
# Buscar la función melt en la documentación oficial
who2 = who.melt(id_vars=["country", "iso2", "iso3", "year"])
who2.head(10)
```

```python
# Crear una columna que contenga el genero (Masculino o femenino)
# Crear una función de Python que contenga las reglas
def asignar_genero(valor):
  if "m" in valor:
    return "masculino"
  else:
    return "femenino"
```

```python
who2["genero"] = who2["variable"].apply(asignar_genero)
who2.head(15)
```

```python
# the method of diagnosis (rel = relapse, sn = negative pulmonary smear, sp = positive pulmonary smear, ep = extrapulmonary)
def asignar_metodo(valor):
  if "rel" in valor:
    return "relapse"
  elif "sn" in valor:
    return "negative pulmonary smear"
  elif "sp" in valor:
    return "positive pulmonary smear"
  elif "ep" in valor:
    return "extrapulmonary"
  else:
    return "to define"
```

```python
who2["metodo"] = who2["variable"].apply(asignar_metodo)
who2.head()
```

```python
# age group (014 = 0-14 yrs of age, 1524 = 15-24, 2534 = 25-34, 3544 = 35-44 years of age, 4554 = 45-54, 5564 = 55-64, 65 = 65 years or older).
def asignar_grupoedad(valor):
  if "014" in valor:
    return "0-14"
  elif "1524" in valor:
    return "15-24"
  elif "2534" in valor:
    return "25-34"
  elif "3544" in valor:
    return "35-44"
  elif "4554" in valor:
    return "45-54"
  elif "5564" in valor:
    return "55-64"
  else:
    return "65+"
```

```python
who2["grupoedad"] = who2["variable"].apply(asignar_grupoedad)
who2.head()
```

```python
# Revelar la columna variable
who2.drop(columns=["variable"], inplace=True)
```

```python
# Reproducir la columna value a casos
who2.rename(columns={"value":"cases"}, inplace=True)
```

```python
who2.head()
```

```python
# Traer el dato de la poblacion total
df = pd.merge(who2, pop, on=["country", "year"], how="inner")
df.head(10)
```

```python
# Nueva tabla para calcular la tasa de incidencia POR AÑO
by_year=df.groupby("year", as_index=False).agg(
    {"cases":"sum", "population":"sum"}
)
by_year["tasa_incidencia"]= 100_000*by_year["cases"]/by_year["population"]
by_year
```

```python
import matplotlib.pyplot as plt #base de visualización en Python
import seaborn as sns # visualizaciones mas sencillas hechas sobre Matplotlib
import plotly.express as px # visualizaciones interactivas
```

```python
import os # Import the os module
# Crear nuestra primera visualización con seaborn
plt.figure(figsize=(10,6))
sns.barplot(x="year", y="tasa_incidencia", data=by_year)
plt.title("Tasa de incidencia por año")
plt.xlabel("Año")
plt.ylabel("Tasa de incidencia")
# guardar la grafica en nuestro drive
output_dir = "/content/drive/MyDrive/data/" # Corrected path to MyDrive
os.makedirs(output_dir, exist_ok=True)
plt.savefig(os.path.join(output_dir, "grafica_incidencia.png"))
```
